This notebook analyzes the statistical distributions of lexical, structural, and host features across benign websites and zero-day quishing attacks.

```python
import pandas as pd
import numpy as np
from qrshield.url_layer.train import build_benchmark_dataset
from qrshield.url_layer.features import extract_url_features, FEATURE_ORDER

raw_data = build_benchmark_dataset()
print(f"Total benchmark samples: {len(raw_data)}")

rows = []
for item in raw_data:
    feats, _, _ = extract_url_features(item['url'], offline=True)
    feats['label'] = item['label']
    rows.append(feats)

df = pd.DataFrame(rows)
df.head()
```

### Statistical Comparison: Benign vs Quishing

```python
grouped = df.groupby('label')[['url_length', 'host_entropy', 'is_abused_tld', 'is_typosquat', 'domain_age_days']].mean()
grouped.index = ['Benign (0)', 'Quishing (1)']
print("Mean feature values across classes:")
print(grouped)
```
